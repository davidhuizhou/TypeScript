### An Example Node.js Application


This code first includes the Node.js [http module](https://nodejs.org/api/http.html).

Node.js has a fantastic [standard library](https://nodejs.org/api/), including first-class support for networking.

```typescript
const { createServer } = require('node:http');

const hostname = '127.0.0.1';
const port = 3000;

const server = createServer((req, res) => {
  res.statusCode = 200;
  res.setHeader('Content-Type', 'text/plain');
  res.end('Hello World');
});

server.listen(port, hostname, () => {
  console.log(`Server running at http://${hostname}:${port}/`);
});
```

### How To Stop

To stop a Node.js HTTP server started with tslab in a VS Code Jupyter Notebook, you can use one of the following methods, depending on whether you want to stop just the server or the entire notebook execution.

1. Stop the Kernel (Most Reliable)
The most direct way to kill a running server and free its port is to restart or stop the Jupyter kernel. This terminates the Node.js process managed by tslab. [1] 

* Restart Kernel: Click the Restart button (circular arrow icon) in the notebook toolbar at the top of the editor.
* Interrupt Kernel: Click the Interrupt button (square stop icon) in the toolbar. This may stop the current execution but might not always release the network port if the server loop is deeply nested. [2, 3, 4] 


Force Kill via Terminal [7] 

If the port remains "in use" after closing the notebook, you can manually kill the process from the VS Code terminal. [8, 9] 
On macOS/Linux:

   1. Find the Process ID (PID) using the port (e.g., 3000):
   lsof -i :3000
   2. Kill the process:
   kill -9 <PID> [10, 11] 

On Windows (PowerShell):

   1. Find the PID:
   netstat -ano | findstr :3000
   2. Kill the process:
   stop-process -Id <PID> -Force [8, 12] 

4. Close the Notebook Tab [11] 
In VS Code, closing the notebook tab or the editor window usually shuts down the local Jupyter server and its associated kernels, though this behavior can vary based on your [VS Code Jupyter extension settings](https://code.visualstudio.com/docs/datascience/jupyter-notebooks). [1, 13, 14, 15]
